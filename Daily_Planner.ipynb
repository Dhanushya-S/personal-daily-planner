```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

# Load environment variables
load_dotenv(override=True)

# Connect to Groq AI model
llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)

# Send a test message
response = llm.invoke(
    "Hello! Introduce yourself in two sentences."
)

# Display AI response
print("Groq Connection Successful!")
print("\nAI Response:")
print(response.content)
```

```python
from datetime import datetime
from zoneinfo import ZoneInfo
from langchain_core.tools import tool

@tool
def get_current_datetime() -> str:
    """Get the current date, time and day in Indian Standard Time."""

    now = datetime.now(ZoneInfo("Asia/Kolkata"))

    return now.strftime(
        "Date: %d-%m-%Y, Time: %I:%M %p, Day: %A"
    )

print("Date and Time Tool Created Successfully!")
print(get_current_datetime.invoke({}))
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

# Create short-term memory
memory = InMemorySaver()

# Define the AI agent's personality
system_prompt = """
You are a Personal Daily Planner Assistant.

Your responsibilities:
1. Help users organize their daily tasks.
2. Prepare realistic daily schedules.
3. Consider task priorities and deadlines.
4. Remember tasks from previous conversations.
5. Modify schedules when users request changes.
6. Answer questions about previously shared tasks.

Rules:
- Use the get_current_datetime tool when current date or time is needed.
- Never invent task details.
- Give clear and practical schedules.
- Remember previous conversation information.
- Be friendly and concise.
"""

# Register the date and time tool
tools = [get_current_datetime]

# Create the single AI agent
planner_agent = create_agent(
    model=llm,
    tools=tools,
    system_prompt=system_prompt,
    checkpointer=memory
)

print("Personal Daily Planner Agent Created Successfully!")
print("Agent Type: Single AI Agent")
print("Memory: InMemorySaver")
print("Tool: Date and Time Tool")
```

```python
# Configure the conversation thread
config = {
    "configurable": {
        "thread_id": "daily-planner-001"
    },
    "run_name": "Personal Daily Planner Demo",
    "tags": ["project"]
}

print("Conversation thread configured!")
print("Thread ID:", config["configurable"]["thread_id"])
```

```python
# TURN 1: User shares daily tasks

turn1 = planner_agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "I have Java study, assignment and exercise today. Prepare my schedule. I have only 4 hours."
            }
        ]
    },
    config=config
)

print("TURN 1 - DAILY SCHEDULE")
print(turn1["messages"][-1].content)
```

```python
# TURN 2: Modify the existing schedule

turn2 = planner_agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Move Java study to evening. Keep my assignment and exercise in the schedule."
            }
        ]
    },
    config=config
)

print("TURN 2 - UPDATED SCHEDULE")
print(turn2["messages"][-1].content)
```

```python
# TURN 3: Test conversation memory

turn3 = planner_agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "What was my assignment task?"
            }
        ]
    },
    config=config
)

print("TURN 3 - MEMORY TEST")
print(turn3["messages"][-1].content)
```

```python
# Follow-up memory verification

memory_test = planner_agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "In the schedule you created earlier, how much time did you allocate for assignment work, and what did you suggest I do?"
            }
        ]
    },
    config=config
)

print("MEMORY VERIFICATION")
print(memory_test["messages"][-1].content)
```

```python
from agent import create_daily_schedule

schedule = create_daily_schedule(
    tasks="""
    Python Study - High Priority
    Java Practice - High Priority
    Exercise - Medium Priority
    """,
    available_hours=4,
    start_time="04:00 PM",
    instructions="Give more time for Python Study",
    date="07 November 2026"
)

print("LANGGRAPH AGENT TEST")
print(schedule)
```

```python
import json

schedule_data = json.loads(schedule)

print("SCHEDULE VALIDATION")
print("Total Activities:", len(schedule_data["schedule"]))

for item in schedule_data["schedule"]:
    print(
        item["activity"],
        "|", item["start_time"],
        "-", item["end_time"],
        "|", item["duration_minutes"], "minutes"
    )

total_minutes = sum(
    item["duration_minutes"]
    for item in schedule_data["schedule"]
)

print("\nTotal Duration:", total_minutes, "minutes")
print("Available Duration: 240 minutes")
print("Within 4 Hours:", total_minutes <= 240)
```

```python
from datetime import datetime

items = schedule_data["schedule"]
overlaps = []

for i in range(1, len(items)):
    previous_end = datetime.strptime(
        items[i - 1]["end_time"], "%I:%M %p"
    )
    current_start = datetime.strptime(
        items[i]["start_time"], "%I:%M %p"
    )

    if current_start < previous_end:
        overlaps.append(items[i]["activity"])

print("OVERLAP CHECK:", "PASS" if not overlaps else "FAIL")
print("Overlapping Activities:", overlaps)
```
